```python
GOAL_STATE = (1, 2, 3, 4, 5, 6, 7, 8, 0)
INITIAL_STATE = (1, 2, 3, 0, 4, 6, 7, 5, 8)

def get_neighbors(state):
    neighbors = []
    idx = state.index(0)
    row, col = idx // 3, idx % 3
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr, nc = row + dr, col + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_idx = nr * 3 + nc
            lst = list(state)
            lst[idx], lst[new_idx] = lst[new_idx], lst[idx]
            neighbors.append(tuple(lst))
    return neighbors

def print_row_of_grids(states, label=""):
    """Prints multiple 3x3 grids horizontally in a single row"""
    if not states: return
    print(label)
    # Loop through the 3 rows of the grid matrix
    for r in range(3):
        row_str = []
        for state in states:
            grid_row = [str(x) if x != 0 else '_' for x in state[r*3:(r+1)*3]]
            row_str.append(" ".join(grid_row))
        print("    |    ".join(row_str))
    print()

def explore_by_levels(initial_state, goal_state):
    visited = {initial_state}
    current_level = [initial_state]
    level = 0

    while current_level:
        print(f"\nLEVEL {level}")
        next_level = []

        for state in current_level:
            print_row_of_grids([state], "Parent:")

            branches = get_neighbors(state)
            valid_branches = [b for b in branches if b not in visited]

            print_row_of_grids(valid_branches, "Possibilities:")
            print("-" * 35)

            for b in valid_branches:
                visited.add(b)
                next_level.append(b)
                if b == goal_state:
                    print("\nGOAL REACHED!")
                    print_row_of_grids([goal_state])
                    return

        current_level = next_level
        level += 1

if __name__ == "__main__":
    explore_by_levels(INITIAL_STATE, GOAL_STATE)
```
