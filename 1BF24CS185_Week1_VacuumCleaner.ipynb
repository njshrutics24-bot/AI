```python
#N J SHRUTI _ 1BF24CS185
# Simple reflex agent

class VacuumEnvironment:
    def __init__(self, room_a, room_b, initial_location):
        self.locations = {'A': room_a, 'B': room_b}
        self.agent_location = initial_location

    def get_percept(self):
        return self.agent_location, self.locations[self.agent_location]

    def execute_action(self, action):
        if action == 'Suck':
            self.locations[self.agent_location] = 'Clean'
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'

    def is_completely_clean(self):
        return self.locations['A'] == 'Clean' and self.locations['B'] == 'Clean'

    def display(self):
        print(f"  Rooms: [A: {self.locations['A']}, B: {self.locations['B']}] | Vacuum: Room {self.agent_location}")


def simple_reflex_agent(percept):
    location, status = percept
    if status == 'Dirty':
        return 'Suck'
    elif location == 'A':
        return 'Right'
    elif location == 'B':
        return 'Left'


if __name__ == "__main__":
    # --- SAMPLE INPUT ---
    ROOM_A_START = 'Dirty'
    ROOM_B_START = 'Dirty'
    STARTING_ROOM = 'A'

    print("=== RUNNING: SIMPLE REFLEX AGENT ===")
    env = VacuumEnvironment(ROOM_A_START, ROOM_B_START, STARTING_ROOM)
    env.display()

    steps = 0
    # Runs until an external environment check stops it
    while not env.is_completely_clean() and steps < 5:
        steps += 1
        percept = env.get_percept()
        action = simple_reflex_agent(percept)
        print(f"Step {steps} | Percept: {percept} -> Action: {action}")
        env.execute_action(action)
        env.display()
```

```python
# Goal based agent

class VacuumEnvironment:
    def __init__(self, room_a, room_b, initial_location):
        self.locations = {'A': room_a, 'B': room_b}
        self.agent_location = initial_location

    def get_percept(self):
        return self.agent_location, self.locations[self.agent_location]

    def execute_action(self, action):
        if action == 'Suck':
            self.locations[self.agent_location] = 'Clean'
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'

    def display(self):
        print(f"  Rooms: [A: {self.locations['A']}, B: {self.locations['B']}] | Vacuum: Room {self.agent_location}")


class GoalBasedVacuumAgent:
    def __init__(self):
        self.goal = {'A': 'Clean', 'B': 'Clean'}
        self.model = {'A': 'Unknown', 'B': 'Unknown'}

    def act(self, percept):
        location, status = percept
        self.model[location] = status  # Update memory tracker

        # Goal check
        if self.model['A'] == self.goal['A'] and self.model['B'] == self.goal['B']:
            return 'NoOp'

        if status == 'Dirty':
            return 'Suck'

        if location == 'A' and self.model['B'] != 'Clean':
            return 'Right'
        elif location == 'B' and self.model['A'] != 'Clean':
            return 'Left'

        return 'NoOp'


if __name__ == "__main__":
    # --- SAMPLE INPUT ---
    ROOM_A_START = 'Dirty'
    ROOM_B_START = 'Dirty'
    STARTING_ROOM = 'A'

    print("=== RUNNING: GOAL-BASED AGENT ===")
    env = VacuumEnvironment(ROOM_A_START, ROOM_B_START, STARTING_ROOM)
    agent = GoalBasedVacuumAgent()
    env.display()

    steps = 0
    while steps < 5:
        steps += 1
        percept = env.get_percept()
        action = agent.act(percept)
        print(f"Step {steps} | Percept: {percept} -> Action: {action}")

        if action == 'NoOp':
            print("Goal achieved! Agent stopped voluntarily.")
            break

        env.execute_action(action)
        env.display()
```
